## 2) 如何部署：`langgraph dev`

1. 在仓库根目录准备好 `.env`（参考 `.env.example`），以及本 `langgraph.json`；
2. 执行 `uv sync` 安装依赖；
3. 在项目根目录运行：

   ```bash
   langgraph dev
   ```

   它会读取 `langgraph.json`，把其中 `graphs` 里声明的图（如 `my_agent`）以 API 形式暴露出来，
   并给出本地访问地址与 Studio 可视化界面；
4. 生产部署可把它推到 LangGraph Platform 或自托管（配置方式同源，只是运行环境不同）。

> 关键点：
> - `dependencies` 指向 `pyproject.toml`（含 jupyter/langgraph 等）；
> - `graphs` 的键值 `"my_agent": "./app/simpleGraph.py:workflow"` 表示把 `simpleGraph.py` 里的
>   `workflow` 对象暴露成名为 `my_agent` 的图；
> - `env` 指向 `.env`，部署时由运行环境注入密钥。

## 3) `main.py` 只是占位入口
